```python
emp_casted = emp_red.withColumn("salary", col("salary").cast("double"))
```

```python
emp_taxed = emp_casted.withColumn("tax", col("salary")*0.2)
```

```python
emp_taxed.show()
```

```python
from pyspark.sql.functions import lit
emp_new_cols = emp_taxed.withColumn("columnOne", lit(1)).withColumn("columnTwo", lit(2))
```

```python
emp_new_cols .show()
```

```python
emp_new_cols.withColumnRenamed("employee_id", "emp_id").show()
```

```python
emp_new_cols.select(expr("employee_id as emp_id"), "name", "age", col("salary"), "tax", emp_new_cols.columnOne, "columnTwo").show()
```

```python
emp_droped = emp_new_cols.drop("columnOne", "columnTwo")
emp_droped.show()
```
