```python
from pyspark.sql import SparkSession

spark = (
    SparkSession
    .builder
    .appName("Sort Union & Aggregation")
    .master("local[*]")
    .getOrCreate()
)

spark
```

```python
emp_data_1 = spark.createDataFrame(data=emp_data_1, schema=emp_schema)
emp_data_2 = spark.createDataFrame(data=emp_data_2, schema=emp_schema)
```

```python
emp_data_1.show()
emp_data_2.show()
```

```python
emp_data_1.printSchema()
emp_data_2.printSchema()
```

```python
# UNION and UNION ALL

emp = emp_data_1.unionAll(emp_data_2)
emp.show()

emp.printSchema()
```

```python
from pyspark.sql.functions import cast, col
from pyspark.sql.types import IntegerType
emp = emp.withColumn("salary", col("salary").cast(IntegerType()))

# emp = emp.withColumn("salary", emp["salary"].cast(IntegerType()))
# ou encore:
# emp = emp.withColumn("salary", col("salary").cast("int"))
```

```python
emp.printSchema()
```

```python
#sort by order respect to salary
from pyspark.sql.functions import desc, asc
#emp_sorted = emp.orderBy("salary")
emp_sorted = emp.orderBy(col("salary").asc())
emp_sorted.show()
```

```python
#desc
from pyspark.sql.functions import desc, asc, count

emp_sorted = emp.orderBy(col("salary").desc())
emp_sorted.show()
```

```python
# select dept_id, count(employee_id) as total_dept_count from emp_sorted group by dept_id 
# emp_count = emp_sorted.groupBy("department_id").agg(count("employee_id").alias("total_dep_count"))

emp_count = emp_sorted.groupBy("department_id").count().withColumnRenamed("count", "total_dep_count")
emp_count.show()
```

```python
from pyspark.sql.functions import countDistinct

emp_count = (
    emp_sorted
    .groupBy("department_id")
    .agg(countDistinct("employee_id").alias("total_dep_count"))
)
emp_count.show()
```

```python
emp_sorted.printSchema()

# type(emp_sorted['salary'])
```

```python
from pyspark.sql.functions import sum, avg
emp_salary_sum = emp_sorted.groupBy("department_id").agg(sum("salary").alias("total_dept_Salary"))

emp_salary_sum.show()
```

```python
emp_avg = emp_sorted.groupBy("department_id").agg(avg("salary").alias("avg_dept_salary")).where("avg_dept_salary>50000")

emp_avg.show()
```
